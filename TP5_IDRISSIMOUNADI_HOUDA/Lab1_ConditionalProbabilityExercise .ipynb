# Conditional Probability Activity & Exercise

Below is some code to create some fake data on how much stuff people purchase given their age range.

It generates 100,000 random "people" and randomly assigns them as being in their 20's, 30's, 40's, 50's, 60's, or 70's.

It then assigns a lower probability for young people to buy stuff.

In the end, we have two Python dictionaries:

"totals" contains the total number of people in each age group.
"purchases" contains the total number of things purchased by people in each age group.
The grand total of purchases is in totalPurchases, and we know the total number of people is 100,000.

Let's run it and have a look:

```python
from numpy import random
random.seed(0)

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0
for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    purchaseProbability = float(ageDecade) / 100.0
    totals[ageDecade] += 1
    if (random.random() < purchaseProbability):
        totalPurchases += 1
        purchases[ageDecade] += 1
```

Let's play with conditional probability.

First let's compute P(E|F), where E is "purchase" and F is "you're in your 30's". The probability of someone in their 30's buying something is just the percentage of how many 30-year-olds bought something:

```python
PEF = float(purchases[30]) / float(totals[30])
print('P(purchase | 30s): ' + str(PEF))
```

P(F) is just the probability of being 30 in this data set:

```python
PF = float(totals[30]) / 100000.0
print("P(30's): " +  str(PF))
```

And P(E) is the overall probability of buying something, regardless of your age:

```python
PE = float(totalPurchases) / 100000.0
print("P(Purchase):" + str(PE))
```

```python
print("P(30's, Purchase)" + str(float(purchases[30]) / 100000.0))
```

Let's also compute the product of P(E) and P(F), P(E)P(F):

```python
print("P(30's)P(Purchase)" + str(PE * PF))
```

Are E and F dependent or independent?

We can also compute P(E|F) = P(E,F)/P(F):

```python
print((purchases[30] / 100000.0) / PF)
```

Modify the code above such that the purchase probability does NOT vary with age, making E and F actually independent.

Then, confirm that P(E|F) is about the same as P(E), showing that the conditional probability of purchase for a given age is not any different than the a-priori probability of purchase regardless of age.


### Réponse à la question du cours : E et F sont-ils dépendants ?

Avec les données d'origine, la probabilité d'achat croît avec l'âge (`age/100`). On compare :

- `P(E)·P(F)` (valeur attendue **si** E et F étaient indépendants)
- `P(E, F)` (probabilité conjointe réelle)

Ces deux valeurs sont différentes (≈ 0.0745 contre ≈ 0.0500 pour les 30 ans), et `P(E|F)` ≈ 0.30 ≠ `P(E)` ≈ 0.45. **E et F sont donc dépendants** : connaître l'âge change la probabilité d'achat.

## Solution de l'exercice

On modifie la simulation : la probabilité d'achat est maintenant **constante** (ici 0.45, indépendante de l'âge). Ainsi E (« achat ») et F (« avoir la trentaine ») deviennent indépendants et on doit retrouver `P(E|F) ≈ P(E)`.

```python
from numpy import random
random.seed(0)

totals2 = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases2 = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases2 = 0
N = 100000
FIXED_PROB = 0.45          # <-- ne dépend plus de l'âge

for _ in range(N):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    purchaseProbability = FIXED_PROB      # avant : float(ageDecade) / 100.0
    totals2[ageDecade] += 1
    if random.random() < purchaseProbability:
        totalPurchases2 += 1
        purchases2[ageDecade] += 1

print("totals    :", totals2)
print("purchases :", purchases2)
print("total achats :", totalPurchases2)
```

Calcul de `P(E|F)`, `P(F)`, `P(E)`, `P(E,F)` et `P(E)P(F)` pour les 30 ans :

```python
PEF2 = purchases2[30] / totals2[30]          # P(E|F)
PF2  = totals2[30] / N                       # P(F)
PE2  = totalPurchases2 / N                   # P(E)
PEandF2 = purchases2[30] / N                 # P(E,F)

print(f"P(purchase | 30s) = {PEF2:.4f}")
print(f"P(purchase)       = {PE2:.4f}")
print(f"P(30's)           = {PF2:.4f}")
print(f"P(30's, purchase) = {PEandF2:.4f}")
print(f"P(30's)P(purchase)= {PE2*PF2:.4f}")
print(f"|P(E|F) - P(E)|   = {abs(PEF2-PE2):.4f}")
```

Vérification pour **toutes** les tranches d'âge :

```python
print("Tranche | P(E|F)  | P(E)   | écart")
for age in totals2:
    p = purchases2[age] / totals2[age]
    print(f"  {age}s   | {p:.4f}  | {PE2:.4f} | {abs(p-PE2):.4f}")
```

### Conclusion

Quand la probabilité d'achat ne dépend plus de l'âge, on obtient `P(E|F) ≈ P(E)` (≈ 0.45 dans les deux cas, l'écart restant de l'ordre de la fluctuation d'échantillonnage) et `P(E,F) ≈ P(E)P(F)`. La probabilité conditionnelle d'achat pour un âge donné n'est donc pas différente de la probabilité a priori : **E et F sont indépendants**, ce qui est exactement la définition `P(E|F) = P(E)`.
