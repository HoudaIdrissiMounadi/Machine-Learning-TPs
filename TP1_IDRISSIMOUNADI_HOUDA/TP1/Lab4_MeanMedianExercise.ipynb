# Exercise: Mean & Median Customer Spend

Here's some code that will generate some random e-commerce data; just an array of total amount spent per transaction. Select the code block, and hit "play" to execute it:

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 20.0, 10000)

plt.hist(incomes, 50)
plt.show()
```

Now, find the mean and median of this data. In the code block below, write your code, and see if your result makes sense:

```python
print("Moyenne :", np.mean(incomes))
print("Médiane :", np.median(incomes))
```

Les deux valeurs sont très proches de 100 (la moyenne qu'on a donnée à `np.random.normal`). C'est normal car la distribution est symétrique (gaussienne), donc moyenne et médiane se confondent.

This is pretty much the world's easiest assignment, but we're just trying to get your hands on iPython and writing code with numpy to get you comfortable with it.

Try playing with the code above to generate different distributions of data, or add outliers to it to see their effect.

```python
# on rajoute quelques valeurs aberrantes pour voir l'effet
incomes2 = np.append(incomes, [5000, 8000, 10000])

print("Sans outliers : moyenne =", round(np.mean(incomes), 2), " médiane =", round(np.median(incomes), 2))
print("Avec outliers : moyenne =", round(np.mean(incomes2), 2), " médiane =", round(np.median(incomes2), 2))

# autre distribution pour tester
autre = np.random.normal(300.0, 80.0, 10000)
print("Autre distribution : moyenne =", round(np.mean(autre), 2), " médiane =", round(np.median(autre), 2))
```

Avec 3 valeurs aberrantes la moyenne augmente nettement alors que la médiane ne bouge presque pas. La médiane est donc plus robuste aux outliers que la moyenne.
