# Standard Deviation and Variance

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 50.0, 10000)

plt.hist(incomes, 50)
plt.show()
```

```python
incomes.std()
```

```python
incomes.var()
```

## Activity

Experiment with different parameters on the normal function, and see what effect it has on the shape of the distribution. How does that new shape relate to the standard deviation and variance?

```python
# on garde la même moyenne (100) et on change l'écart-type
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharex=True)
for ax, sigma in zip(axes, [10.0, 50.0, 100.0]):
    data = np.random.normal(100.0, sigma, 10000)
    ax.hist(data, 50)
    ax.set_title("sigma = " + str(sigma))
    print("sigma =", sigma, "-> std =", round(data.std(), 2), " var =", round(data.var(), 2))
plt.show()
```

```python
# maintenant on change la moyenne en gardant sigma = 50
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharex=True)
for ax, mu in zip(axes, [0.0, 100.0, 200.0]):
    data = np.random.normal(mu, 50.0, 10000)
    ax.hist(data, 50)
    ax.set_title("mu = " + str(mu))
    print("mu =", mu, "-> std =", round(data.std(), 2), " var =", round(data.var(), 2))
plt.show()
```

Quand on augmente l'écart-type la courbe devient plus large et plus aplatie (les valeurs sont plus dispersées autour de la moyenne), et quand il est petit la courbe est étroite et haute. La variance est le carré de l'écart-type donc elle grandit encore plus vite (10 -> 100, 50 -> 2500, 100 -> 10000).

Par contre, changer la moyenne ne fait que déplacer la courbe vers la gauche ou la droite : la forme, l'écart-type et la variance restent à peu près les mêmes.
