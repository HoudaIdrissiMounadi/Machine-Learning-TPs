## Functions

```python
def SquareIt(x):
    return x * x

print(SquareIt(2))
```

```python
#You can pass functions around as parameters
def DoSomething(f, x):
    return f(x)

print(DoSomething(SquareIt, 3))
```

```python
#Lambda functions let you inline simple functions
#Python lambda functions, also known as anonymous functions, are inline functions that do not have a name. 
#They are created with the lambda keyword.
print(DoSomething(lambda x: x * x * x, 3))
```

Write a Python program to split a given list into two parts where the length of the first part of the list is given. 
Original list:
[1, 1, 2, 3, 4, 4, 5, 1]
Length of the first part of the list: 3
Splited the said list into two parts:
([1, 1, 2], [3, 4, 4, 5, 1])

```python
liste = [1, 1, 2, 3, 4, 4, 5, 1]
n = 3

print("Original list:")
print(liste)
print("Length of the first part of the list:", n)

partie1 = liste[:n]
partie2 = liste[n:]
print("Splited the said list into two parts:")
print((partie1, partie2))
```

Write a Python program to find the item with maximum occurrences in a given list.
Original list:
[2, 3, 8, 4, 7, 9, 8, 2, 6, 5, 1, 6, 1, 2, 3, 4, 6, 9, 1, 2]
Item with maximum occurrences of the said list:
2

```python
liste = [2, 3, 8, 4, 7, 9, 8, 2, 6, 5, 1, 6, 1, 2, 3, 4, 6, 9, 1, 2]

# on compte les occurrences de chaque élément avec un dictionnaire
compteur = {}
for e in liste:
    if e in compteur:
        compteur[e] += 1
    else:
        compteur[e] = 1
print(compteur)

# l'élément qui a le plus grand nombre d'occurrences
maxi = 0
for e in compteur:
    if compteur[e] > maxi:
        maxi = compteur[e]
        element = e

print("Original list:")
print(liste)
print("Item with maximum occurrences of the said list:")
print(element, "(", maxi, "fois )")

# verification rapide avec max()
print(max(liste, key=liste.count))
```
